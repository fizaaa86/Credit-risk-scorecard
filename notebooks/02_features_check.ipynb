```python
import duckdb
con = duckdb.connect()
con.sql("""
    SELECT SK_ID_CURR, CREDIT_ACTIVE, AMT_CREDIT_SUM, AMT_CREDIT_SUM_DEBT, CREDIT_DAY_OVERDUE
    FROM '../data/parquet/bureau.parquet'
    WHERE SK_ID_CURR = 215354
""")
```

```python
con.sql("""
    SELECT NAME_CONTRACT_STATUS, AMT_APPLICATION, AMT_CREDIT, DAYS_DECISION
    FROM '../data/parquet/previous_application.parquet'
    WHERE SK_ID_CURR = 215354
    ORDER BY DAYS_DECISION
""")
```

```python
con.sql("""
    SELECT SK_ID_PREV, NUM_INSTALMENT_NUMBER, DAYS_INSTALMENT, DAYS_ENTRY_PAYMENT,
           AMT_INSTALMENT, AMT_PAYMENT
    FROM '../data/parquet/installments_payments.parquet'
    WHERE SK_ID_CURR = 215354
    ORDER BY DAYS_INSTALMENT DESC
    LIMIT 15
""")
```
